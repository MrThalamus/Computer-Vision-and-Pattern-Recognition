```python
# -*- coding: utf-8 -*-

# =========================================================
# SOROKH-POTH
# VGG16 FINE-TUNED RCNN STYLE
# CLASSIFICATION + BBOX REGRESSION + GRAD-CAM
# =========================================================

import os
import cv2
import glob
import time
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms, models

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    mean_squared_error,
    mean_absolute_error
)

# =========================================================
# DEVICE
# =========================================================

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("DEVICE:", DEVICE)

# =========================================================
# DATASET PATHS
# =========================================================
# =========================================================
# SOROKH-POTH DATASET PATHS
# =========================================================

BASE_PATH = "/kaggle/input/datasets/fahadhassanfahim162/vehicle-data/Sorokh-Poth A Balanced Bangladeshi Road Vehicle Image Dataset/Sorokh-Poth A Balanced Bangladeshi Road Vehicle Image Dataset"

DATASET_INFO = {

    "AutoRickshaw": {
        "img": BASE_PATH + "/AutoRickshaw_Image-20230408T190450Z-001/AutoRickshaw_Image",
        "ann": BASE_PATH + "/AutoRickshaw_Annotations-20230408T190450Z-001/AutoRickshaw_Annotations"
    },

    "BiCycle": {
        "img": BASE_PATH + "/BiCycle_Image-20230408T174520Z-001/BiCycle_Image",
        "ann": BASE_PATH + "/BiCycle_Annotations-20230408T151922Z-001/BiCycle_Annotations"
    },

    "Bike": {
        "img": BASE_PATH + "/Bike_Image-20230408T175508Z-001/Bike_Image",
        "ann": BASE_PATH + "/Bike_Annotations-20230408T175502Z-001/Bike_Annotations"
    },

    "Bus": {
        "img": BASE_PATH + "/Bus_Image-20230408T175800Z-001/Bus_Image",
        "ann": BASE_PATH + "/Bus_Annotations-20230408T175758Z-001/Bus_Annotations"
    },

    "CNG": {
        "img": BASE_PATH + "/CNG_Image-20230408T181646Z-001/CNG_Image",
        "ann": BASE_PATH + "/CNG_Annotations-20230408T181642Z-001/CNG_Annotations"
    },

    "Car": {
        "img": BASE_PATH + "/Car_Image-20230408T181401Z-001/Car_Image",
        "ann": BASE_PATH + "/Car_Annotations-20230408T181357Z-001/Car_Annotations"
    },

    "Leguna": {
        "img": BASE_PATH + "/Leguna_Image-20230408T182004Z-001/Leguna_Image",
        "ann": BASE_PATH + "/Leguna_Annotations-20230408T182000Z-001/Leguna_Annotations"
    },

    "Rickshaw": {
        "img": BASE_PATH + "/Rickshaw_Image-20230408T182344Z-001/Rickshaw_Image",
        "ann": BASE_PATH + "/Rickshaw_Annotations-20230408T182340Z-001/Rickshaw_Annotations"
    },

    "Truck": {
        "img": BASE_PATH + "/Truck_Image-20230408T182357Z-001/Truck_Image",
        "ann": BASE_PATH + "/Truck_Annotations-20230408T182351Z-001/Truck_Annotations"
    },

    "VAN": {
        "img": BASE_PATH + "/VAN_Images-20230408T185622Z-001/VAN_Images",
        "ann": BASE_PATH + "/VAN_Annotations-20230408T185628Z-001/VAN_Annotations"
    }
}

# =========================================================
# CONFIGURATION
# =========================================================

IMAGE_SIZE = 224
BATCH_SIZE = 32
EPOCHS = 20
LEARNING_RATE = 1e-5

CLASS_NAMES = list(DATASET_INFO.keys())
NUM_CLASSES = len(CLASS_NAMES)

print("Classes:", CLASS_NAMES)
print("Number of classes:", NUM_CLASSES)

# =========================================================
# TRANSFORMS
# =========================================================

train_transform = transforms.Compose([

    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),

    transforms.RandomHorizontalFlip(),

    transforms.RandomRotation(10),

    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

val_transform = transforms.Compose([

    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# =========================================================
# DATASET CLASS
# =========================================================

class BanglaRoadDataset(Dataset):

    def __init__(self, dataset_info, transform=None):

        self.transform = transform

        self.samples = []

        self.class_to_idx = {
            name: idx
            for idx, name in enumerate(dataset_info.keys())
        }

        for class_name, paths in dataset_info.items():

            img_dir = paths["img"]
            ann_dir = paths["ann"]

            class_id = self.class_to_idx[class_name]

            image_paths = []

            for ext in ["*.jpg", "*.jpeg", "*.png", "*.JPG", "*.JPEG", "*.PNG"]:

                image_paths.extend(
                    glob.glob(
                        os.path.join(img_dir, ext)
                    )
                )

            for img_path in image_paths:

                filename = os.path.basename(img_path)

                txt_name = os.path.splitext(filename)[0] + ".txt"

                ann_path = os.path.join(
                    ann_dir,
                    txt_name
                )

                self.samples.append(
                    (
                        img_path,
                        ann_path,
                        class_id
                    )
                )

    def __len__(self):

        return len(self.samples)

    def __getitem__(self, idx):

        image_path, ann_path, class_id = self.samples[idx]

        image = Image.open(
            image_path
        ).convert("RGB")

        bbox = [0.5, 0.5, 0.5, 0.5]

        if os.path.exists(ann_path):

            with open(ann_path, "r") as f:

                lines = f.readlines()

            for line in lines:

                values = line.strip().split()

                if len(values) >= 5:

                    try:

                        bbox = [
                            float(values[1]),
                            float(values[2]),
                            float(values[3]),
                            float(values[4])
                        ]

                        break

                    except:

                        pass

        if self.transform:

            image = self.transform(image)

        return (
            image,
            torch.tensor(class_id, dtype=torch.long),
            torch.tensor(bbox, dtype=torch.float32)
        )

# =========================================================
# DATASET SPLIT
# =========================================================

base_dataset = BanglaRoadDataset(
    DATASET_INFO,
    transform=None
)

total_size = len(base_dataset)

print("Total Images:", total_size)

if total_size == 0:

    raise ValueError("No images found. Check dataset paths.")

train_size = int(0.7 * total_size)

valid_size = int(0.15 * total_size)

test_size = total_size - train_size - valid_size

indices = torch.randperm(
    total_size
).tolist()

train_indices = indices[:train_size]

valid_indices = indices[
    train_size:train_size + valid_size
]

test_indices = indices[
    train_size + valid_size:
]

train_full = BanglaRoadDataset(
    DATASET_INFO,
    transform=train_transform
)

valid_full = BanglaRoadDataset(
    DATASET_INFO,
    transform=val_transform
)

test_full = BanglaRoadDataset(
    DATASET_INFO,
    transform=val_transform
)

train_dataset = Subset(
    train_full,
    train_indices
)

valid_dataset = Subset(
    valid_full,
    valid_indices
)

test_dataset = Subset(
    test_full,
    test_indices
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True
)

valid_loader = DataLoader(
    valid_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)

print("Train:", len(train_dataset))
print("Valid:", len(valid_dataset))
print("Test :", len(test_dataset))

# =========================================================
# VGG16 FINE-TUNED RCNN MODEL
# =========================================================

class VGG16_RCNN_FineTune(nn.Module):

    def __init__(self, num_classes):

        super(VGG16_RCNN_FineTune, self).__init__()

        vgg = models.vgg16(
            weights=models.VGG16_Weights.IMAGENET1K_V1
        )

        self.features = vgg.features

        # Freeze early layers
        for param in self.features[:20].parameters():

            param.requires_grad = False

        # Fine-tune deeper layers
        for param in self.features[20:].parameters():

            param.requires_grad = True

        self.pool = nn.AdaptiveAvgPool2d((7, 7))

        self.shared_fc = nn.Sequential(

            nn.Flatten(),

            nn.Linear(25088, 4096),
            nn.ReLU(),
            nn.Dropout(0.5),

            nn.Linear(4096, 1024),
            nn.ReLU(),
            nn.Dropout(0.5)
        )

        self.classifier = nn.Linear(
            1024,
            num_classes
        )

        self.bbox_regressor = nn.Linear(
            1024,
            4
        )

    def forward(self, x):

        x = self.features(x)

        x = self.pool(x)

        x = self.shared_fc(x)

        class_output = self.classifier(x)

        bbox_output = torch.sigmoid(
            self.bbox_regressor(x)
        )

        return class_output, bbox_output

# =========================================================
# INITIALIZE MODEL
# =========================================================

model = VGG16_RCNN_FineTune(
    NUM_CLASSES
).to(DEVICE)

# =========================================================
# LOSS FUNCTIONS + OPTIMIZER
# =========================================================

classification_loss_fn = nn.CrossEntropyLoss()

bbox_loss_fn = nn.SmoothL1Loss()

optimizer = optim.Adam(
    filter(
        lambda p: p.requires_grad,
        model.parameters()
    ),
    lr=LEARNING_RATE
)

# =========================================================
# TRAIN FUNCTION
# =========================================================

def train_one_epoch(model, loader):

    model.train()

    running_loss = 0
    running_cls_loss = 0
    running_bbox_loss = 0

    correct = 0
    total = 0

    for images, labels, bboxes in tqdm(loader):

        images = images.to(DEVICE)

        labels = labels.to(DEVICE)

        bboxes = bboxes.to(DEVICE)

        optimizer.zero_grad()

        class_outputs, bbox_outputs = model(images)

        classification_loss = classification_loss_fn(
            class_outputs,
            labels
        )

        bbox_loss = bbox_loss_fn(
            bbox_outputs,
            bboxes
        )

        total_loss = classification_loss + bbox_loss

        total_loss.backward()

        optimizer.step()

        running_loss += total_loss.item()
        running_cls_loss += classification_loss.item()
        running_bbox_loss += bbox_loss.item()

        preds = torch.argmax(
            class_outputs,
            dim=1
        )

        correct += (
            preds == labels
        ).sum().item()

        total += labels.size(0)

    return (
        running_loss / len(loader),
        running_cls_loss / len(loader),
        running_bbox_loss / len(loader),
        correct / total
    )

# =========================================================
# EVALUATION FUNCTION
# =========================================================

@torch.no_grad()
def evaluate(model, loader, phase=None):

    model.eval()

    running_loss = 0
    running_cls_loss = 0
    running_bbox_loss = 0

    all_preds = []
    all_labels = []

    all_bbox_true = []
    all_bbox_pred = []

    for images, labels, bboxes in loader:

        images = images.to(DEVICE)

        labels = labels.to(DEVICE)

        bboxes = bboxes.to(DEVICE)

        class_outputs, bbox_outputs = model(images)

        classification_loss = classification_loss_fn(
            class_outputs,
            labels
        )

        bbox_loss = bbox_loss_fn(
            bbox_outputs,
            bboxes
        )

        total_loss = classification_loss + bbox_loss

        running_loss += total_loss.item()
        running_cls_loss += classification_loss.item()
        running_bbox_loss += bbox_loss.item()

        preds = torch.argmax(
            class_outputs,
            dim=1
        )

        all_preds.extend(
            preds.cpu().numpy()
        )

        all_labels.extend(
            labels.cpu().numpy()
        )

        all_bbox_true.extend(
            bboxes.cpu().numpy()
        )

        all_bbox_pred.extend(
            bbox_outputs.cpu().numpy()
        )

    avg_loss = running_loss / len(loader)

    avg_cls_loss = running_cls_loss / len(loader)

    avg_bbox_loss = running_bbox_loss / len(loader)

    accuracy = accuracy_score(
        all_labels,
        all_preds
    )

    precision = precision_score(
        all_labels,
        all_preds,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        all_labels,
        all_preds,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        all_labels,
        all_preds,
        average="weighted",
        zero_division=0
    )

    bbox_mse = mean_squared_error(
        np.array(all_bbox_true),
        np.array(all_bbox_pred)
    )

    bbox_mae = mean_absolute_error(
        np.array(all_bbox_true),
        np.array(all_bbox_pred)
    )

    if phase is not None:

        print("\n")
        print("=" * 60)
        print(f"{phase} RESULTS")
        print("=" * 60)

        print(f"Total Loss      : {avg_loss:.4f}")
        print(f"Class Loss      : {avg_cls_loss:.4f}")
        print(f"BBox Loss       : {avg_bbox_loss:.4f}")
        print(f"Accuracy        : {accuracy:.4f}")
        print(f"Precision       : {precision:.4f}")
        print(f"Recall          : {recall:.4f}")
        print(f"F1 Score        : {f1:.4f}")
        print(f"BBox MSE        : {bbox_mse:.4f}")
        print(f"BBox MAE        : {bbox_mae:.4f}")

        print("\nClassification Report:\n")

        print(
            classification_report(
                all_labels,
                all_preds,
                labels=list(range(NUM_CLASSES)),
                target_names=CLASS_NAMES,
                zero_division=0
            )
        )

        print("=" * 60)

    return (
        avg_loss,
        avg_cls_loss,
        avg_bbox_loss,
        accuracy,
        precision,
        recall,
        f1,
        bbox_mse,
        bbox_mae,
        all_labels,
        all_preds
    )

# =========================================================
# TRAINING LOOP
# =========================================================

start_time = time.time()

train_losses = []
val_losses = []

train_cls_losses = []
val_cls_losses = []

train_bbox_losses = []
val_bbox_losses = []

train_accs = []
val_accs = []

for epoch in range(EPOCHS):

    train_loss, train_cls_loss, train_bbox_loss, train_acc = train_one_epoch(
        model,
        train_loader
    )

    (
        val_loss,
        val_cls_loss,
        val_bbox_loss,
        val_acc,
        _,
        _,
        _,
        _,
        _,
        _,
        _
    ) = evaluate(
        model,
        valid_loader,
        phase=None
    )

    train_losses.append(train_loss)
    val_losses.append(val_loss)

    train_cls_losses.append(train_cls_loss)
    val_cls_losses.append(val_cls_loss)

    train_bbox_losses.append(train_bbox_loss)
    val_bbox_losses.append(val_bbox_loss)

    train_accs.append(train_acc)
    val_accs.append(val_acc)

    print(f"\nEpoch {epoch + 1}/{EPOCHS}")

    print(
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {val_loss:.4f}"
    )

    print(
        f"Train Class Loss: {train_cls_loss:.4f} | "
        f"Val Class Loss: {val_cls_loss:.4f}"
    )

    print(
        f"Train BBox Loss: {train_bbox_loss:.4f} | "
        f"Val BBox Loss: {val_bbox_loss:.4f}"
    )

    print(
        f"Train Acc: {train_acc:.4f} | "
        f"Val Acc: {val_acc:.4f}"
    )

# =========================================================
# SAVE MODEL
# =========================================================

torch.save(
    model.state_dict(),
    "VGG16_FineTuned_RCNN.pth"
)

print("\nSaved model: VGG16_FineTuned_RCNN.pth")

# =========================================================
# FINAL VALIDATION + TEST
# =========================================================

evaluate(
    model,
    valid_loader,
    phase="VALIDATION"
)

(
    test_loss,
    test_cls_loss,
    test_bbox_loss,
    test_acc,
    test_precision,
    test_recall,
    test_f1,
    test_bbox_mse,
    test_bbox_mae,
    y_true,
    y_pred
) = evaluate(
    model,
    test_loader,
    phase="TEST"
)

# =========================================================
# CONFUSION MATRIX
# =========================================================

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=list(range(NUM_CLASSES))
)

plt.figure(figsize=(12, 10))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=CLASS_NAMES,
    yticklabels=CLASS_NAMES
)

plt.title("VGG16 Fine-Tuned Confusion Matrix")

plt.xlabel("Predicted")

plt.ylabel("True")

plt.show()

# =========================================================
# TRAINING CURVES
# =========================================================

epochs_range = range(1, EPOCHS + 1)

plt.figure(figsize=(10, 5))
plt.plot(epochs_range, train_losses, label="Train Loss")
plt.plot(epochs_range, val_losses, label="Validation Loss")
plt.legend()
plt.title("Total Loss Curve")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(epochs_range, train_cls_losses, label="Train Class Loss")
plt.plot(epochs_range, val_cls_losses, label="Validation Class Loss")
plt.legend()
plt.title("Classification Loss Curve")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(epochs_range, train_bbox_losses, label="Train BBox Loss")
plt.plot(epochs_range, val_bbox_losses, label="Validation BBox Loss")
plt.legend()
plt.title("BBox Regression Loss Curve")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(epochs_range, train_accs, label="Train Accuracy")
plt.plot(epochs_range, val_accs, label="Validation Accuracy")
plt.legend()
plt.title("Accuracy Curve")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.show()

# =========================================================
# REMOVE INPLACE RELU FOR GRAD-CAM
# =========================================================

for module in model.modules():

    if isinstance(module, nn.ReLU):

        module.inplace = False

# =========================================================
# GRAD-CAM
# =========================================================

class GradCAM:

    def __init__(self, model, target_layer):

        self.model = model

        self.target_layer = target_layer

        self.gradients = None

        self.activations = None

        self.forward_handle = self.target_layer.register_forward_hook(
            self.forward_hook
        )

        self.backward_handle = self.target_layer.register_full_backward_hook(
            self.backward_hook
        )

    def forward_hook(self, module, input, output):

        self.activations = output

    def backward_hook(self, module, grad_input, grad_output):

        self.gradients = grad_output[0]

    def generate(self, input_tensor, target_class):

        self.gradients = None

        self.activations = None

        self.model.eval()

        input_tensor.requires_grad = True

        class_output, _ = self.model(input_tensor)

        score = class_output[0, target_class]

        self.model.zero_grad()

        score.backward(retain_graph=True)

        if self.gradients is None:

            raise ValueError("Gradient extraction failed.")

        if self.activations is None:

            raise ValueError("Activation extraction failed.")

        gradients = self.gradients[0]

        activations = self.activations[0]

        weights = torch.mean(
            gradients,
            dim=(1, 2)
        )

        cam = torch.zeros(
            activations.shape[1:],
            dtype=torch.float32
        ).to(DEVICE)

        for i, w in enumerate(weights):

            cam += w * activations[i]

        cam = torch.relu(cam)

        cam -= cam.min()

        cam /= cam.max() + 1e-8

        return cam.detach().cpu().numpy()

# =========================================================
# SHOW GRAD-CAM + PREDICTED BBOX
# =========================================================

target_layer = model.features[28]

gradcam = GradCAM(
    model,
    target_layer
)

def show_gradcam(dataset, index=0):

    image_tensor, label, true_bbox = dataset[index]

    input_tensor = image_tensor.unsqueeze(0).to(DEVICE)

    model.eval()

    with torch.enable_grad():

        class_output, bbox_output = model(input_tensor)

        pred_class = torch.argmax(
            class_output,
            dim=1
        ).item()

        pred_bbox = bbox_output[0].detach().cpu().numpy()

        heatmap = gradcam.generate(
            input_tensor,
            pred_class
        )

    image = image_tensor.permute(
        1,
        2,
        0
    ).cpu().numpy()

    mean = np.array([0.485, 0.456, 0.406])

    std = np.array([0.229, 0.224, 0.225])

    image = image * std + mean

    image = np.clip(image, 0, 1)

    image_uint8 = np.uint8(image * 255)

    x_center, y_center, width, height = pred_bbox

    x_center *= IMAGE_SIZE
    y_center *= IMAGE_SIZE

    width *= IMAGE_SIZE
    height *= IMAGE_SIZE

    x1 = int(x_center - width / 2)
    y1 = int(y_center - height / 2)

    x2 = int(x_center + width / 2)
    y2 = int(y_center + height / 2)

    x1 = max(0, min(IMAGE_SIZE - 1, x1))
    y1 = max(0, min(IMAGE_SIZE - 1, y1))
    x2 = max(0, min(IMAGE_SIZE - 1, x2))
    y2 = max(0, min(IMAGE_SIZE - 1, y2))

    bbox_image = image_uint8.copy()

    cv2.rectangle(
        bbox_image,
        (x1, y1),
        (x2, y2),
        (0, 255, 0),
        2
    )

    cv2.putText(
        bbox_image,
        f"Pred: {CLASS_NAMES[pred_class]}",
        (x1, max(20, y1 - 10)),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (0, 255, 0),
        2
    )

    heatmap = cv2.resize(
        heatmap,
        (IMAGE_SIZE, IMAGE_SIZE)
    )

    heatmap = np.uint8(255 * heatmap)

    heatmap = cv2.applyColorMap(
        heatmap,
        cv2.COLORMAP_JET
    )

    superimposed = cv2.addWeighted(
        image_uint8,
        0.6,
        heatmap,
        0.4,
        0
    )

    plt.figure(figsize=(18, 6))

    plt.subplot(1, 3, 1)
    plt.imshow(image)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(
        cv2.cvtColor(
            bbox_image,
            cv2.COLOR_BGR2RGB
        )
    )
    plt.title(
        f"Predicted BBox\nClass: {CLASS_NAMES[pred_class]}"
    )
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(
        cv2.cvtColor(
            superimposed,
            cv2.COLOR_BGR2RGB
        )
    )
    plt.title(
        f"Grad-CAM\nPredicted Class: {CLASS_NAMES[pred_class]}"
    )
    plt.axis("off")

    plt.show()

show_gradcam(
    test_dataset,
    index=1
)

# =========================================================
# FINAL RESULT TABLE
# =========================================================

import pandas as pd

final_result_table = pd.DataFrame({
    "Model": ["VGG16 Fine-Tuned RCNN"],

    "Accuracy": [test_acc],

    "Precision": [test_precision],

    "Recall": [test_recall],

    "F1-score": [test_f1],

    "BBox MSE": [test_bbox_mse],

    "BBox MAE": [test_bbox_mae],

    "Total Loss": [test_loss],

    "Classification Loss": [test_cls_loss],

    "BBox Loss": [test_bbox_loss]
})

display(final_result_table)

# =========================================================
# COMPLETION MESSAGE
# =========================================================

end_time = time.time()

print("\n" + "=" * 70)
print("VGG16 FINE-TUNING COMPLETED SUCCESSFULLY")
print("=" * 70)
print(f"Total Time: {(end_time - start_time) / 60:.2f} minutes")
print("Saved Model: VGG16_FineTuned_RCNN.pth")
print("=" * 70)
```
